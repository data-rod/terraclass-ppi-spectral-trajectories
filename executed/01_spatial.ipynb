## Define eligible positions

A position must retain the same TerraClass class in 2016, 2018, 2020, 2022 and 2024. The 3×3 erosion keeps positions whose eight neighbours also satisfy permanence. Nearest-neighbour reprojection transfers this mask to the 10 m Brazil Data Cube (BDC) grid; the cell centre determines whether a position belongs to the basin.

The basin intersects tiles 029010 and 030010. Their adjacent cells are processed together for each class and date. The basin area is 459.97 km² in the Albers equal-area projection; the original vector attribute is 463.52 km².
