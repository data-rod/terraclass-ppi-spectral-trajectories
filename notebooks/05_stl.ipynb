## Fit the components and calculate their strengths

Classical STL with robust weighting decomposes reflectance as Yₜ = Tₜ + Sₜ + Rₜ: trend, seasonal component and residual. The period is 23 calendar positions, the trend window is 45 and the low-pass window is 25; the local polynomial degree is one. Seasonal windows are 7, 9, 11, 13, 15, 21, 25, 31 and 35.

Sample variances on the originally observed dates define seasonal strength F_S = max(0, 1 − Var(R)/Var(S+R)) and trend strength F_T = max(0, 1 − Var(R)/Var(T+R)). Time interpolation supplies the regular sequence needed for fitting. Interpolated positions are excluded from the strength calculations; all 173 observed dates per series are included.
