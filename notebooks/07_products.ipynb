# Figures and numerical summaries

This notebook generates print figures from a chosen set of tables and displays the corresponding strength table. It uses archived products unless RBC_PRODUCTS_DIR identifies the output of a completed products command.

````python
from pathlib import Path
import os,sys,json
import pandas as pd
from IPython.display import display,SVG
from terraclass_ppi.paths import CODE_ROOT as ROOT,output_root
from terraclass_ppi.data_io import prepare_reference,table_directory,iter_candidate_groups
WORK=Path(os.environ['RBC_OUTPUT']).resolve() if os.environ.get('RBC_OUTPUT') else prepare_reference()
os.environ['RBC_OUTPUT']=str(WORK)
FIGURES=output_root()/'figures'
from terraclass_ppi.spatial import CFG
````

## Select the figure inputs

To derive tables and diagnostics from a reconstructed model, run:

```text
python scripts/run_pipeline.py products --work outputs/complete --source-work outputs/reference --output-dir outputs/rebuilt_products
```

Set RBC_PRODUCTS_DIR to that output path before starting this kernel. The generator below then reads its `dados/` subdirectory and writes PNG, SVG and PDF figures under `outputs/figures/`. With RBC_PRODUCTS_DIR unset, the figure inputs are the canonical archived tables. The archived figure files remain unchanged.

The products command compares four numerical tables with the reference and recalculates diagnostics through the shared scientific functions. Its absolute tolerances are 10⁻⁸ for STL and 10⁻⁷ for classifier scores. The notebook displays the selected product set; it does not refit the classifier or rerun PPI.

````python
import subprocess
PRODUCTS=Path(os.environ['RBC_PRODUCTS_DIR']) if os.environ.get('RBC_PRODUCTS_DIR') else table_directory()
command=[sys.executable,'-m','terraclass_ppi.figures','--output',str(FIGURES)]
if os.environ.get('RBC_PRODUCTS_DIR'):command.extend(['--data',str(PRODUCTS/'dados')])
subprocess.run(command,check=True)
display(pd.read_csv(PRODUCTS/'table_s2_window7.csv'))
display(SVG(filename=str(FIGURES/'figura_5_sensibilidade.svg')))
````

## Check the table and figure

The strength table groups the ten bands and four spectral domains for window 7. Figure 5 compares F_S and F_T across all nine windows. `outputs/notebooks/execution.json` records notebook execution, while `outputs/` records the separate complete reconstruction and product checks.
