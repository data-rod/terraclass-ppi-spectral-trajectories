# Figures and numerical summaries

This notebook generates print figures from a chosen set of tables and displays the corresponding strength table. It uses archived products unless RBC_PRODUCTS_DIR identifies the output of a completed products command.

````python
from pathlib import Path
import os,sys,json
ROOT = next(p for p in [Path.cwd(),*Path.cwd().parents] if (p/'config.json').exists())
sys.path.insert(0,str(ROOT))
WORK = Path(os.environ.get('RBC_OUTPUT',ROOT/'work')).resolve()
os.environ['RBC_OUTPUT']=str(WORK)
import pandas as pd
from IPython.display import display,SVG
from dados import CFG
````

## Select the figure inputs

To derive tables and diagnostics from a reconstructed model, run:

```text
python vigente.py products --work rebuilt-complete --source-work work --output-dir rebuilt-products
```

Set RBC_PRODUCTS_DIR to that output path before starting this kernel. The generator below then reads its `dados/` subdirectory and writes PNG, SVG and PDF figures under `generated/notebook-figures/`. With RBC_PRODUCTS_DIR unset, the figure inputs are the archived tables in `editorial_figures/dados/`. The archived figure files remain unchanged.

The products command compares four numerical tables with the reference and recalculates diagnostics through the shared scientific functions. Its absolute tolerances are 10⁻⁸ for STL and 10⁻⁷ for classifier scores. The notebook displays the selected product set; it does not refit the classifier or rerun PPI.

````python
import subprocess
PRODUCTS=Path(os.environ.get('RBC_PRODUCTS_DIR',WORK/'artigo')).resolve()
DATA=PRODUCTS/'dados' if os.environ.get('RBC_PRODUCTS_DIR') else ROOT/'editorial_figures/dados'
FIGURES=ROOT/'generated/notebook-figures'
subprocess.run([sys.executable,str(ROOT/'editorial_figures/scripts/figuras_3_4_5.py'),
                '--data',str(DATA),'--output',str(FIGURES)],check=True)
display(pd.read_csv(PRODUCTS/'tabela_2.csv'))
display(SVG(filename=str(FIGURES/'figura_5_sensibilidade.svg')))
````
